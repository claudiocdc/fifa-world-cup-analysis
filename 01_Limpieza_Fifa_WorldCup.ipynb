```python
import pandas as pd
url = "https://raw.githubusercontent.com/claudiocdc/fifa-world-cup-analysis/main/WorldCups.csv"
df = pd.read_csv(url)
print(df.head())
print(df.info())
```

```python
#vemos que la asistencia son texto porque tienen el punto, la limpiamos para pasarlo a número
##Ojo! El regex = False sirve por si acaso, por eso lo pongo. En otras versiones de python , quizás lo trata como true y te borra todo, no solo el "."

df['Attendance'] = df['Attendance'].str.replace(".","",regex=False)
print(df.head())
```

```python
#cambiamos a número entero int
df['Attendance'] = df['Attendance'].astype(int)
print(df.info())
```

```python
#¿qué mundial lleno más estadios?
df.sort_values('Attendance',ascending = False).head()
```

```python
#sorprendente que con menos partidos hubiese mas asistencia que el de 2014 que tuvo mas partidos, suponemos que es por los estadios americanos
##vamos a ver realmente si fue por eso

df['Avg_Attendance'] = df['Attendance'] / df['MatchesPlayed']
df.sort_values('Avg_Attendance',ascending = False).head()
```

```python
#¿Qué país ha ganado más Mundiales en la historia?

df['Winner'].value_counts()
```

```python
#Lo de Alemania es un poco raro, habría que cambiarlo

df['Winner'] = df['Winner'].replace('Germany FR','Germany')
df['Winner'].value_counts()
```

```python
#Vamos a guardar la  tabla de partidos en otro dataframe

df_matches = pd.read_csv('https://raw.githubusercontent.com/claudiocdc/fifa-world-cup-analysis/refs/heads/main/WorldCupMatches.csv')
print(df_matches.head())
print(df_matches.info())
```

```python
#attendance tiene menos, ¿Por qué? y hay muchas filas vacias ya que pone que hay 4572 entries cuando realmente hay solo rellenas 852
##Hay una función muy buena que es .dropna() https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.dropna.html
###Segñun la docu el how ='all' permite borrar las que estan enteramente vacias y no solo las que tengan alguna vacía

df_matches = df_matches.dropna(how='all')
print(df_matches.info())
```

```python
#¿Cuantos mundiales ha jugado cada país?
##Hay que juntar las veces que aparece tanto local como visitante para ver sus apariciones, luego usar la variable de año como indicador para contar cuantas veces aparecieron
###Aparte year esta en float y queda feo, vamos a cambiarlo a int
####Al ejecutar el código más adelante, vemos que hay nombres en sucio con valores como rn">Bosnia and Herzegovina	porlo que quitamos ese 'rn">'

df_matches["Home Team Name"] = df_matches["Home Team Name"].str.replace('rn">', '', regex=False)
df_matches["Away Team Name"] = df_matches["Away Team Name"].str.replace('rn">', '', regex=False)

df_matches['Year'] = df_matches['Year'].astype(int)
locales = df_matches[['Year','Home Team Name']].rename(columns={'Home Team Name':'Equipo'})
visitantes = df_matches[['Year','Away Team Name']].rename(columns={'Away Team Name':'Equipo'})
apariciones = pd.concat([locales, visitantes])
print(apariciones)
```

```python
#Contamos las apariciones por año: agrupar por equipo, y de cada grupo cuántos años únicos tiene. Ojo con la n en el unique
##Confirmamos que Brazil es el que más mundiales ha ganado, también poruqe es el que más mundiales ha jugado

apariciones.groupby('Equipo')['Year'].nunique().sort_values(ascending=False)

```

```python
# Consulta rápida: ¿cuántas veces apareció España?

consulta_pais = 'Spain'
filas_spain = apariciones[apariciones['Equipo'] == consulta_pais]
mundiales = filas_spain['Year'].nunique()
print('El país', consulta_pais, 'ha jugado', mundiales, 'Mundiales hasta 2014')
```
