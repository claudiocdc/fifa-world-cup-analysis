```python
import pandas as pd
url = "https://raw.githubusercontent.com/claudiocdc/fifa-world-cup-analysis/main/WorldCups.csv"
df = pd.read_csv(url)
print(df.head())
print(df.info())
```

```python
#vemos que la asistencia son texto porque tienen el punto, la limpiamos para pasarlo a número


df['Attendance'] = df['Attendance'].str.replace(".","")
print(df.head())
```

```python
#cambiamos a número entero int
df['Attendance'] = df['Attendance'].astype(int)
print(df.info())
```

**¿Qué mundial lleno más estadios?**

```python
df.sort_values('Attendance',ascending = False).head()
```

El Mundial con mayor asistencia total de la historia fue USA 1994, con más de 3,58 millones de espectadores.

```python
#sorprendente que con menos partidos hubiese mas asistencia que el de 2014 que tuvo mas partidos, suponemos que es por los estadios americanos
##vamos a ver realmente si fue por eso

df['Avg_Attendance'] = df['Attendance'] / df['MatchesPlayed']
df.sort_values('Avg_Attendance',ascending = False).head()
```

 USA 1994 lidera también en asistencia media por partido (~68.991), por delante de Mundiales más modernos con más partidos. Confirma que el récord se debe a los grandes estadios estadounidenses, no solo al número de encuentros.

```python
#Vamos a guardar la  tabla de partidos en otro dataframe

df_matches = pd.read_csv('https://raw.githubusercontent.com/claudiocdc/fifa-world-cup-analysis/refs/heads/main/WorldCupMatches.csv')
df_matches = df_matches.drop_duplicates()
print(df_matches.head())
print(df_matches.info())
```

```python
#attendance tiene menos, ¿Por qué? y hay muchas filas vacias ya que pone que hay 4572 entries cuando realmente hay solo rellenas 852
##Hay una función muy buena que es .dropna() https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.dropna.html
###Segñun la docu el how ='all' permite borrar las que estan enteramente vacias y no solo las que tengan alguna vacía

df_matches = df_matches.dropna(how='all')

print(df_matches.info())
```

**¿Cuántos mundiales ha jugado cada país?**

```python
#Hay que juntar las veces que aparece tanto local como visitante para ver sus apariciones, luego usar la variable de año como indicador para contar cuantas veces aparecieron
##Aparte year esta en float y queda feo, vamos a cambiarlo a int
###Al ejecutar el código más adelante, vemos que hay nombres en sucio con valores como rn">Bosnia and Herzegovina	porlo que quitamos ese 'rn">'

df_matches["Home Team Name"] = df_matches["Home Team Name"].str.replace('rn">', '')
df_matches["Away Team Name"] = df_matches["Away Team Name"].str.replace('rn">', '')
df_matches['Home Team Name'] = df_matches['Home Team Name'].replace({'Germany FR':'Germany', 'IR Iran':'Iran'})
df_matches['Away Team Name'] = df_matches['Away Team Name'].replace({'Germany FR':'Germany', 'IR Iran':'Iran'})

df_matches['Year'] = df_matches['Year'].astype(int)
home = df_matches[['Year','Home Team Name']].rename(columns={'Home Team Name':'Team'})
away = df_matches[['Year','Away Team Name']].rename(columns={'Away Team Name':'Team'})
appearances = pd.concat([home, away])
appearances
```

```python
#Contamos las apariciones por año: agrupar por equipo, y de cada grupo cuántos años únicos tiene. Ojo con la n en el unique
##Confirmamos que Brazil es el que más mundiales ha ganado, también poruqe es el que más mundiales ha jugado

appearances.groupby('Team')['Year'].nunique().sort_values(ascending=False)

```

Brasil es la única selección que ha jugado las 20 ediciones (1930–2014). Le siguen Alemania empatada con Italia (18)

**¿Cuántas veces apareció España?**

```python
# Consulta rápida:

consult_country = 'Spain'
rows_spain = appearances[appearances['Team'] == consult_country]
World_cups = rows_spain['Year'].nunique()
print('El país', consult_country, 'ha jugado', World_cups, 'Mundiales hasta 2014')
```

**¿Cuál ha sido el partido con más goles de toda la historia de los Mundiales?**

```python
#Creamos columna con el total de goles y luego ordenamos
##Voy a pasar ambas columnas a int ya que no puedes marcar un gol y medio
###Creamos la variable y ordenamos

df_matches[['Away Team Goals','Home Team Goals']] = df_matches[['Away Team Goals','Home Team Goals']].astype(int)
df_matches['total_goals'] = df_matches['Away Team Goals']+ df_matches['Home Team Goals']
df_matches.sort_values('total_goals',ascending = False).head(1)

```

Austria 7–5 Suiza (1954), con 12 goles.

**¿Qué selección ha metido más goles en toda la historia de los Mundiales?**

```python
#Vamos a crear una tabla con los equipos y los goles tanto de local como de visitante
##Renombramos los nombres para unificarlos https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.rename.html#pandas.DataFrame.rename
###Concatenamos y creamos una tabla con ello
####Finalmente agrupamos los equipos por la suma de goles

df_home_goals = df_matches[['Home Team Name','Home Team Goals']].rename(columns={'Home Team Name':'Team', 'Home Team Goals':'Goals'})
df_away_goals = df_matches[['Away Team Name','Away Team Goals']].rename(columns={'Away Team Name':'Team', 'Away Team Goals':'Goals'})

df_top_goals = pd.concat([df_home_goals, df_away_goals])

df_top_goals.groupby('Team')['Goals'].sum().sort_values(ascending=False)
```

Alemania es la máxima goleadora de la historia de los Mundiales, por delante de Brasil (221 goles). (Ojo: este resultado solo emerge tras unificar "Germany FR" y "Germany"; sin esa limpieza, Brasil parecía el número uno).

**Análisis de jugadores**

```python
df_players = pd.read_csv('https://raw.githubusercontent.com/claudiocdc/fifa-world-cup-analysis/refs/heads/main/WorldCupPlayers.csv')
# Unificamos FRG (Alemania Occidental) con GER: es la misma federación.
# Problema detectado al analizar las tarjetas (ver sección de tarjetas más abajo).
# Se corrige aquí, en el origen, para que se aplique a TODOS los análisis.
df_players['Team Initials'] = df_players['Team Initials'].replace('FRG','GER')
df_players = df_players.drop_duplicates()
print(df_players.info())
```

Nota: un NaN no siempre es "dato sucio". A veces significa "no aplica" o "no ocurrió". En este caso es normal que haya jugadores que no registren eventos (goles,tarjetas...) y el position se entiende que porque se rellena si eres capitan por ejemplo


```python
df_players['Position'].value_counts()
```

Vemos que Position registra marca los roles de portero (goalkeeper) y capitan (captain)

**¿Cuál es el número de camiseta más usado en la historia?**

```python
df_players['Shirt Number'].value_counts()
```

Es un poco raro lo del 0, no sabemos si porque registra a los jugadores que no jugaron con el dorsal 0 o porque antiguamente no usaban dorsal.

```python
#Sabemos que necesitamos que las dos columnas clave para unir dos dataframes tienen que tener mismo formato, sino da error. Igualamos el dtype de la columna de ambas tablas
df_matches['MatchID'] = df_matches['MatchID'].astype(int)

##Unimos por año de partidos a la tabla de jugadores
df_players_year = df_players.merge(df_matches[['MatchID','Year']], on='MatchID', how='left')
print(df_players_year.head())
```

```python
#Ahora vamos a ver cuantos 0 habian por año para ver si es la hipotesis de los mundiales sin dorsales u otra cosa
row_dorsal_0 = df_players_year[df_players_year['Shirt Number'] == 0]
row_dorsal_0['Year'].value_counts()

```

Los dorsales no se usaron en los Mundiales hasta 1954; antes, los jugadores no llevaban número.

**¿Cuál sería el número más usado si quitamos los ceros?**

```python
row_dorsal = df_players_year[df_players_year['Shirt Number'] != 0]
row_dorsal['Shirt Number'].value_counts()
```

Excluyendo el "0" (la era sin dorsales, hasta 1950), el número más usado en la historia es el 1, el dorsal del portero.

**¿Qué dorsal mete más goles?**

```python
#Debemos pillar solo los que contengan "G" = Gol, na=False https://pandas.pydata.org/docs/reference/api/pandas.Series.str.contains.html

goals = df_players_year[df_players_year['Event'].str.contains('G', na=False)]
goals['Shirt Number'].value_counts()
```

Como ha contado los que son True realmente son partidos en los que ese dorsal a metido gol. El 0 por lo de la época que se jugaba sin dorsal, el siguiente sería el 9 (dorsal que usan los delanteros)

```python
#Creamos nueva columna para que cuente los goles
df_players_year['goals'] = df_players_year['Event'].str.count('G')

goals_by_shirt_number = df_players_year.groupby('Shirt Number')['goals'].sum().sort_values(ascending=False)
goals_by_shirt_number
```

Ahora si sabemos que el dorsal 9 es el dorsal con más goles, 256 (quitando el sin dorsal)

**Nota:** *El conteo suma cada "G" del texto, incluyendo algún gol en propia (OG) puntual, así que puede haber una desviación mínima. El dorsal 9 lidera con claridad de todos modos.*

**¿Cúal es la selección con más tarjetas de la historia?**

```python
#Primero hacemos uso de regex para que pille solo las letras de Event y me lo ponga en una tabla, asi comprobamos que tenemos que filtrar
##G = gol, Y = amarilla, R = roja, RSY = roja por doble amarilla
nomenclatura_eventos = df_players['Event'].str.findall(r'[A-Z]+').explode().value_counts()
nomenclatura_eventos
```

Al analizar las tarjetas por selección, detecté que Alemania aparecía partida en dos siglas: FRG (Alemania Occidental) y GER (reunificada), lo que dividía su recuento. Como es la misma federación, la unifiqué en la limpieza inicial (arriba) para que el arreglo se aplicara a todo el análisis.

```python
df_players['Team Initials'].unique()
```

```python
#Sabemos que Y será Yellow y R Red, aparte RSY será "Red for Second Yellow"
df_players_year['yellow_cards'] = df_players_year['Event'].str.count('Y')
df_players_year['red_cards'] = df_players_year['Event'].str.count('R')

df_players_year['total_cards'] = df_players_year['yellow_cards'] + df_players_year['red_cards']

total_cards = df_players_year.groupby('Team Initials')['total_cards'].sum().sort_values(ascending=False)
total_cards.head(10)
```

 Argentina y Alemania son las selecciones con más tarjetas de la historia (123 cada una, hasta 2014). Curiosamente, Alemania solo alcanza lo más alto tras unificar sus siglas FRG (Alemania Occidental) y GER; sin esa limpieza aparecía 6ª con 80.

Nota: el código RSY (roja por doble amarilla) se cuenta como amarilla y como roja a la vez, lo que introduce una desviación mínima (solo 51 casos en toda la historia).

**Correlaciones**

**¿El fútbol se ha vuelto más aburrido?**

 (goles por partido a lo largo de la historia)

```python
#Creamos columna en el dataframe de los mundiales
##Reflejamos la tabla
df['goals_per_match'] = df['GoalsScored']/df['MatchesPlayed']
df[['Year','goals_per_match']]
```

```python
#Vamos a ver dicha relación importando matplotlib
##plt.plot(x, y, marker='o') https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html

import matplotlib.pyplot as plt

plt.plot(df['Year'], df['goals_per_match'], marker='o')
plt.title('¿El fútbol se ha vuelto más aburrido? Goles por partido por Mundial')
plt.xlabel('Año')
plt.ylabel('Goles por partido')
plt.show()
```

```python
#Hacemos correlación para verificar que correlación existe
##https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.corr.html

df['Year'].corr(df['goals_per_match'])
```

Existe una correlación negativa fuerte (-0,75) entre el año y los goles por partido: el fútbol mete menos goles hoy que en los años 50 (pico en 1954).


Esto no significa que los años sean la causa sino de una evolución posiblemente táctica del juego en los mundiales.

**¿Las selecciones que más goles marcan son también las más tarjeteras?**

```python
#Necesitamos una tabla con goles y tarjetas por selección, juntos
##Ya tenemos una con goles y tarjetas solo hay que agrupar y sumar por equipos
###Ojeamos por verificar que se ha creado bien y hacemos la correlación

team_stats = df_players_year.groupby('Team Initials')[['goals','total_cards']].sum()
team_stats.sort_values('goals',ascending = False)
team_stats['goals'].corr(team_stats['total_cards'])
```

De primeras podemos decir que existe una correlación fuerte con respecto a que a más goles más tarjetas, aún así es un poco injusto comentar esto ya que los equipos que llegan más lejos, tendrán más estadisticas, no es que se meta un gol y te echen una amarilla...
**Normalicemos sacando goles por partido y tarjetas por partido, quitando así la variable número de partidos**

```python
df_players_year.head()
```

```python
#Creamos columna de partidos unicos por equipo
team_stats['matches'] = df_players_year.groupby('Team Initials')['MatchID'].nunique()
team_stats['matches'].head()
```

```python
#Creamos las dos columnas con su variable
team_stats['goals_per_match'] = team_stats['goals']/team_stats['matches']
team_stats['card_per_match'] = team_stats['total_cards']/team_stats['matches']
team_stats['goals_per_match'].corr(team_stats['card_per_match'])
```

```python
team_stats.head()
```

 La correlación bruta entre goles y tarjetas por selección era fuerte (0,84), pero al normalizar por partidos jugados cae a -0,30. La relación aparente estaba causada por los partidos disputados: los equipos que llegan lejos acumulan más de todo. "Los goleadores son los más sucios" resulta ser un ilusión estadística.
