```python
import pandas as pd
url = "https://raw.githubusercontent.com/claudiocdc/fifa-world-cup-analysis/main/WorldCups.csv"
df = pd.read_csv(url)
print(df.head())
print(df.info())
```

```python
#vemos que la asistencia son texto porque tienen el punto, la limpiamos para pasarlo a número
##Ojo! El regex = False sirve por si acaso, por eso lo pongo. En otras versiones de python , quizás lo trata como true y te borra todo, no solo el "."

df['Attendance'] = df['Attendance'].str.replace(".","",regex=False)
print(df.head())
```

```python
#cambiamos a número entero int
df['Attendance'] = df['Attendance'].astype(int)
print(df.info())
```

```python
#¿qué mundial lleno más estadios?
df.sort_values('Attendance',ascending = False).head()
```

```python
#sorprendente que con menos partidos hubiese mas asistencia que el de 2014 que tuvo mas partidos, suponemos que es por los estadios americanos
##vamos a ver realmente si fue por eso

df['Avg_Attendance'] = df['Attendance'] / df['MatchesPlayed']
df.sort_values('Avg_Attendance',ascending = False).head()
```

```python
#¿Qué país ha ganado más Mundiales en la historia?

df['Winner'].value_counts()
```

```python
#Lo de Alemania es un poco raro, habría que cambiarlo

df['Winner'] = df['Winner'].replace('Germany FR','Germany')
df['Winner'].value_counts()
```
